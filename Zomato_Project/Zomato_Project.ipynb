```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## step 2 : Create Data Frame

```python
dataframe = pd.read_csv("Zomato data .csv")
print()
```

```python
dataframe
```

```python
## Convert the data type of Column - rate
```

```python
def handleRate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)

dataframe['rate'] = dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

## Type of Resturants

```python
dataframe.head()
```

```python
sns.countplot(x = dataframe['listed_in(type)'])
plt.xlabel("Type of Resturant")
```

##### Conclusion : majority of the rseturant falls in dinning category

#### Ranking of Resturants 

```python
dataframe.head()
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c="green", marker="o")
plt.xlabel("Type of Resturant", c = "red", size = 20)
plt.ylabel("votes", c = "red", size = 20)
```

### Conclusion : Dining resturants received maximum votes

```python
plt.hist(dataframe['rate'], bins = 5)
plt.title("Rating Distribution")
plt.show()
```

### Conclusion : Majority resturants received ratings from 3.5 to 4 

### Avg order spending by couples

```python
couple_data = dataframe['approx_cost(for two people)']
sns.countplot(x = couple_data)
```

### Conclusion : The majority of couples prefer restrunts with an approx cost of 300

### Offline or Online which mode received max ratings 

```python
dataframe.head()
```

```python
plt.figure(figsize = (6,6))
sns.boxplot(x = 'online_order', y = 'rate', data = dataframe)
```

### Conclusion : Peoples those who are ordering online they are giving more ratings than offline

```python
dataframe.head()
```

```python
pivot_table = dataframe.pivot_table(index = 'listed_in(type)', columns = 'online_order', aggfunc = 'size', fill_value = 0)
sns.heatmap(pivot_table, annot = True, cmap = "YlGnBu", fmt = 'd')
plt.title("Heatmap")
plt.xlabel("Online_Order")
plt.ylabel("Listed In (Type)")
plt.show()
```

### Conclusion : Dining restrunts primarily accepts offline orders, whereas cafes primarily receives online orders. This suggest that clients prefers orders in person at resturants, but prefer online ordering at cafes
